### Combine all months of 2024


```python
from pathlib import Path

# Locate the project data directory from either the notebook or project folder.
data_dir = Path("data") if Path("data").is_dir() else Path("../data")
df_jan = pd.read_parquet(data_dir / "yellow_taxi_2024_01_aggreg.parquet")

df_2024 = pd.concat(
    [df_jan, df_feb, df_mar, df_apr, df_may, df_jun,
     df_jul, df_aug, df_sep, df_oct, df_nov, df_dec],
    ignore_index = True,
).sort_values(["tpep_pickup_datetime", "PULocationID"]).reset_index(drop = True)

# Every hour of the leap year should contain all 263 zones.
assert len(df_2024) == 366 * 24 * 263
assert not df_2024.duplicated(["tpep_pickup_datetime", "PULocationID"]).any()
assert df_2024["tpep_pickup_datetime"].min() == pd.Timestamp("2024-01-01")
assert df_2024["tpep_pickup_datetime"].max() == pd.Timestamp("2024-12-31 23:00:00")

df_2024.to_parquet(data_dir / "yellow_taxi_2024_aggreg.parquet", index = False)
df_2024
```
